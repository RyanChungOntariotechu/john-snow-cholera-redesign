```python
from PIL import Image
import plotly.graph_objects as go
import pandas as pd
import plotly.io as pio
import plotly.express as px
from pyproj import Transformer
from sklearn.neighbors import KNeighborsClassifier
import numpy as np
import os
```

```python
#Load the data
pio.renderers.default = "notebook"
deathData=pd.read_csv("deaths.csv")
pumpData=pd.read_csv("pumps.csv")
snowDates=pd.read_csv("snow_dates.csv")
```

```python
#Save charts
os.makedirs("../charts", exist_ok=True)

FONT_CSS = """<style>
@import url('https://fonts.googleapis.com/css2?family=Redaction:ital,wght@0,400;0,700;1,400&display=swap');
</style>"""

def save_chart(fig, name):
    path = f"../charts/{name}.html"
    fig.write_html(path, include_plotlyjs="directory", config={"responsive": True})
    html = open(path, encoding="utf-8").read()
    html = html.replace("<head>", "<head>" + FONT_CSS, 1)
    open(path, "w", encoding="utf-8").write(html)
```

```python
#Standardize GPS coordinates to meters
transformer = Transformer.from_crs("EPSG:4326", "EPSG:32630", always_xy=True)

broad = pumpData.loc[pumpData["pump_name"] == "Broad St."].iloc[0]
#set origin to Broad St
origin_x, origin_y = transformer.transform(broad["y_coordinate"], broad["x_coordinate"])

def convert_meters(lon, lat):
    '''
    1. Takes two parameters lon and lat 
    2. Converts them to plain array
    3. Converts degrees to meters
    4. Subtracts by the Broad St pump to its own position (Broad St. is origin)
    '''
    x, y = transformer.transform(lon.to_numpy(), lat.to_numpy())
    return (x - origin_x), (y - origin_y)
#Using convert_meters I convert deathData and pumpData GPS coordinates to meters
deathData["x_meters"], deathData["y_meters"] = convert_meters(deathData["y_longitude"], deathData["x_latitude"])
pumpData["x_meters"],  pumpData["y_meters"]  = convert_meters(pumpData["y_coordinate"], pumpData["x_coordinate"])
```

```python
deathData.head()
```

```python
pumpData.head()
```

```python
#Combine deaths in same location 
deathsGrouped = (deathData.groupby(["x_meters", "y_meters", "x_latitude", "y_longitude"],
                                    as_index=False)["death_count"].sum())
```

```python
#Create a model that finds the nearest neighbor for each death
model = KNeighborsClassifier(n_neighbors=1)
model.fit(pumpData[["x_meters", "y_meters"]], pumpData["pump_name"])

#Create a column for nearest pump based on the model I just created
deathsGrouped["nearest_pump"] = model.predict(deathsGrouped[["x_meters", "y_meters"]])

```

# Map Chart

This chart was made to replicate and upgrade the original chart, giving a better visual for the audience.
The map chart with the radius off shows us deaths closest to each pump by colour.
With the radius on, it counts deaths within the radius of Broad St.
## Rationale: 
- I needed a way to check the distance of deaths from each pump
- Radius OFF: Using the KNeighborsClassifier, it can tell me which deaths are closest to each pump
- Radius ON: Using the Radius

## Conclusion:
- By far, the deaths were closest to the Broad St pump
- However this doesn't fully confirm that deaths strictly came from Broad St
- Timeline chart is also needed to tell me how removing the pump helped 

```python
#Takes colors from the plotly express library
colors = px.colors.qualitative.Bold

#Maps pump names to corresponding colors
color_map = dict(zip(pumpData["pump_name"], colors))

mapFig = go.Figure()

#Deaths colored by their nearest pump
mapFig.add_trace(go.Scattermap(
    lat=deathsGrouped["x_latitude"], 
    lon=deathsGrouped["y_longitude"],
    mode="markers",
    name="Deaths",
    customdata=np.c_[deathsGrouped["death_count"], deathsGrouped["nearest_pump"]],
    hovertemplate=("<b>Deaths here:</b> %{customdata[0]}<br>"
                   "<b>Nearest pump:</b> %{customdata[1]}<extra></extra>"),
    marker=dict(size=6 + deathsGrouped["death_count"] * 3,
                color=deathsGrouped["nearest_pump"].map(color_map), opacity=0.5),
    showlegend=False
))

#Pumps
mapFig.add_trace(go.Scattermap(
    lat=pumpData["x_coordinate"],
    lon=pumpData["y_coordinate"],
    mode="markers+text", name="Pumps",
    text=pumpData["pump_name"], textposition="top center",
    marker=dict(size=16, color=pumpData["pump_name"].map(color_map)),
    showlegend=False
))

# Radius circle around Broad St. (Stays empty until moved)
mapFig.add_trace(go.Scattermap(lat=[],
                            lon=[],
                            mode="lines", 
                            name="Radius",
                            showlegend=False,
                            line=dict(color="crimson", width=3), 
                            hoverinfo="skip"))

# Uses a color list to give out distinct colors for each pump
for pump, color in color_map.items():
    mapFig.add_trace(go.Scattermap(lat=[None],
                                lon=[None],
                                mode="markers",
                                name=pump,
                                marker=dict(size=12,
                                            color=color)))

'''----------------------------Radius Slider---------------------------- '''

broad = pumpData.loc[pumpData["pump_name"] == "Broad St."].iloc[0]
#dist (np.hypot) calculates 2d distance from the origin 
dist = np.hypot(deathsGrouped["x_meters"], deathsGrouped["y_meters"])
#Select the broad x and y coordinate
lat0 = broad["x_coordinate"]
lon0 = broad["y_coordinate"]
#Angle ensures that the circle is even all around
angle = np.linspace(0, 2 * np.pi, 100)
#Gives me a range from 0 to 300 in 25 intervals ( 0 = back to original colors)
radii = [0] + list(range(25, 301, 25)) 
#Resets colors back to original
original = deathsGrouped["nearest_pump"].map(color_map)

def radius_slider(radius):
    '''
    1. Inside is a true or false statement: Every death location inside or equal to radius counts true
    2. go.frame builds the slider
        i. Labels frame on the slider for the name
        ii.Converts radius from meters to degress of latitude for scatterplot
        iii.Changes title based on radius and adds up death count inside radius
    '''
    inside = dist <= radius
    return go.Frame(
        name=str(radius), traces=[0, 2],
        data=[go.Scattermap(marker=dict(color=np.where(inside, "crimson", "lightgray") if radius else original)),
              go.Scattermap(lat=lat0 + (radius / 111320) * np.sin(angle),
                            lon=lon0 + (radius / (111320 * np.cos(np.radians(lat0)))) * np.cos(angle))],
        layout=go.Layout(title_text=f"Within {radius} m of Broad St.: {deathsGrouped['death_count'][inside].sum()} deaths"
                                    if radius else "Cholera Deaths by Nearest Water Pump: Soho, 1854"))

mapFig.frames = [radius_slider(radius) for radius in radii]

mapFig.update_layout(
    title=dict(text="Cholera Deaths by Nearest Water Pump: Soho, 1854",
               x=0.5, xanchor="center", y=0.97, yanchor="top"),
    map=dict(style="carto-positron", center=dict(lat=51.5133, lon=-0.1367), zoom=15.5),
    paper_bgcolor="black",
    plot_bgcolor="black",
    font=dict(family="Redaction, Georgia, serif", color="#8ecae6"),
    width=None,
    height=800,
    autosize=True,
    legend=dict(title=dict(text="Pumps", font=dict(color="#8ecae6")),
                font=dict(color="#8ecae6")),
    sliders=[dict(bgcolor="#8ecae6",
                  currentvalue=dict(prefix="Radius (m): "), pad=dict(t=30), steps=[
        dict(method="animate", label="Off" if radius == 0 else str(radius),
             args=[[str(radius)], {"frame": {"duration": 0, "redraw": True}, "mode": "immediate"}])
        for radius in radii])],
)
mapFig.show()
save_chart(mapFig, "map")
```

# Chapter Timeline Chart:

Shows me a timeline of both cases and deaths together.
Rationale:
- I can see whether removing the pump helped in Chapter 4
- Tells me the peak of cases and when cases started to dissipate

Conclusion:
- Cases were already dropping before the handle was removed 
- Tells me that removing the handle didn't end the outbreak; rather, it probably stopped a 2nd resurgence
Notes: I noticed that both the snowdates and the other two data weren't synced together, so I separated it into a different chart

```python
snowDates["date"] = pd.to_datetime(snowDates["date"])
snowDates.rename(columns={'attacks': 'cases'}, inplace=True)
```

```python
timelineChapters = [
    ("1. A quiet start", "1854-08-19", "1854-08-29", "Cases start to appear: 7 deaths in the first 11 days."),
    ("2. The spark",     "1854-08-30", "1854-08-31", "Cases surge: 56 new cases on 31 Aug alone."),
    ("3. The explosion", "1854-09-01", "1854-09-03", "The peak: 143 cases on 1 Sept, 127 deaths on 2 Sept."),
    ("4. The handle",    "1854-09-08", "1854-09-08", "Handle removed, but new cases had already fallen from 143 to 12."),
    ("5. The tail",      "1854-09-09", "1854-09-30", "Fading out: 1 new case by 12 Sept, no deaths by 30 Sept."),
]

# labels at each peak replace the legend (kept in every chapter, since relayout replaces annotations)
labels = [dict(x="1854-09-01", 
               y=143, text="Cases", 
               xanchor="right", 
               xshift=-8,
               showarrow=False, 
               font=dict(color='#12436D', size=14)),
          dict(x="1854-09-02", y=127, text="Deaths", 
               xanchor="left", 
               xshift=8,
               showarrow=False, 
               font=dict(color='#F46A25', 
                         size=14))]

def chapter_layout(start, end, text):
    '''
    1. Takes a chapter start, end and caption
    2. Every time you move slider, it moves a yellow box (showing a specific timeline)
    3. Returns a layout dictionary with cases and death labels. (Relayout replaces all annotations, so the labels must be used)
    '''
    shape = dict(type="rect", xref="x", yref="paper", y0=0, y1=1,
                 x0=pd.Timestamp(start) - pd.Timedelta(hours=12),
                 x1=pd.Timestamp(end) + pd.Timedelta(hours=12),
                 fillcolor="#575757", opacity=0.5, line_width=0, layer="below")
    note = dict(x=0, y=1.14, xref="paper", yref="paper", xanchor="left",
                text=f"<b>{text}</b>", showarrow=False, font=dict(size=20, color='#8ecae6'))
    return {"shapes": [shape], "annotations": [note] + labels}

timelineFig = go.Figure()
timelineFig.add_trace(go.Scatter(x=snowDates["date"],
                         y=snowDates["cases"],
                         name="Cases per day",
                         mode="lines", 
                         line=dict(color='#12436D', width=3),
                         fill="tozeroy", 
                         fillcolor="rgba(143,163,184,0.35)"))

timelineFig.add_trace(go.Scatter(x=snowDates["date"],
                         y=snowDates["deaths"], 
                         name="Deaths per day",
                         mode="lines", 
                         line=dict(color='#F46A25', width=3),
                         fill="tozeroy", fillcolor="rgba(192,57,43,0.35)"))
#Builds the layout for chapter 1
first = chapter_layout(*timelineChapters[0][1:])

timelineFig.update_layout(
    font=dict(family="Redaction,Georgia", color='#8ecae6'),
    paper_bgcolor="black",
    plot_bgcolor="black",
    title=dict(
        text="Cholera Outbreak Timeline: Soho, 1854",
        x=0.5,            
        xanchor="center", 
        y=0.97,           
        yanchor="top",
    ),
    xaxis=dict(tickformat="%d %b", showgrid=False, title=None),
    yaxis=dict(gridcolor="#e5e8ec", rangemode="tozero", title="Per day"),
    shapes=first["shapes"], annotations=first["annotations"],
    margin=dict(t=150, b=120, r=60),
    #Builds slider with chapters
    sliders=[dict(currentvalue=dict(prefix="Chapter: ", font=dict(size=16)),
                  pad=dict(t=50),
                  bgcolor="#8ecae6",
                  activebgcolor='#F46A25',
                  bordercolor="#e5e8ec",
                  steps=[dict(method="relayout",
                              label=name,
                              args=[chapter_layout(start, end, text)])
                         for name, start, end, text in timelineChapters])],height=700)
timelineFig.show()
save_chart(timelineFig, "timeline")
```

# Stats for both charts

```python
#Map chart
total = deathsGrouped["death_count"].sum()
broad_deaths = deathsGrouped.loc[deathsGrouped["nearest_pump"] == "Broad St.", "death_count"].sum()

print(f"{broad_deaths} of {total} deaths ({broad_deaths / total:.1%}) were closest to Broad St.")
```

```python
#Radius Chart
total = deathsGrouped["death_count"].sum()

# Tells me to iterate from 100m,150m,250m from Broad St Pump and outputs percentage
radiusStats = []
for radius in (100, 150, 250):
    deathMeter = int(deathsGrouped.loc[dist <= radius, "death_count"].sum())
    radiusStats.append((radius, deathMeter, deathMeter / total))

# Deaths whose nearest pump is Broad St.
deathBroad = int(deathsGrouped.loc[deathsGrouped["nearest_pump"] == "Broad St.", "death_count"].sum())

print(f"Total deaths: {int(total)}")
for radius, deathMeter, share in radiusStats:
    print(f"Within {radius} m: {deathMeter} ({share:.0%})")
print(f"Nearest pump is Broad St.: {deathBroad} ({deathBroad / total:.0%})")
```
