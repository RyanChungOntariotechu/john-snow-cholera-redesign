# John Snow Cholera Redesign: Version 2

**What changed from v1:** v1 only plotted the deaths and pumps on a map. In v2 I measure the data and make the map interactive.

- Convert GPS coordinates to meters, with the Broad St. pump as the origin
- Use a nearest-neighbor model to find the closest pump for every death
- Colour each death by its nearest pump
- Add a radius slider that counts the deaths around Broad St.
- Add a first, simple timeline of cases and deaths

**Not in this version yet (added in v3):** chapter slider on the timeline, stats, dark styling and the website.

```python
import plotly.graph_objects as go
import pandas as pd
import plotly.io as pio
import plotly.express as px
from pyproj import Transformer
from sklearn.neighbors import KNeighborsClassifier
import numpy as np
```

```python
#Load the data
pio.renderers.default = "notebook"
deathData=pd.read_csv("deaths.csv")
pumpData=pd.read_csv("pumps.csv")
snowDates=pd.read_csv("snow_dates.csv")
```

```python
#Standardize GPS coordinates to meters
transformer = Transformer.from_crs("EPSG:4326", "EPSG:32630", always_xy=True)

broad = pumpData.loc[pumpData["pump_name"] == "Broad St."].iloc[0]
#set origin to Broad St
origin_x, origin_y = transformer.transform(broad["y_coordinate"], broad["x_coordinate"])

def convert_meters(lon, lat):
    '''
    1. Takes two parameters lon and lat
    2. Converts them to plain arrays
    3. Converts degrees to meters
    4. Subtracts the Broad St. position so that Broad St. is the origin (0, 0)
    '''
    x, y = transformer.transform(lon.to_numpy(), lat.to_numpy())
    return (x - origin_x), (y - origin_y)

deathData["x_meters"], deathData["y_meters"] = convert_meters(deathData["y_longitude"], deathData["x_latitude"])
pumpData["x_meters"],  pumpData["y_meters"]  = convert_meters(pumpData["y_coordinate"], pumpData["x_coordinate"])
```

```python
#Combine deaths in same location
deaths_grouped = (deathData.groupby(["x_meters", "y_meters", "x_latitude", "y_longitude"],
                                    as_index=False)["death_count"].sum())

#Create a model that finds the nearest pump for each death location
model = KNeighborsClassifier(n_neighbors=1)
model.fit(pumpData[["x_meters", "y_meters"]], pumpData["pump_name"])
deaths_grouped["nearest_pump"] = model.predict(deaths_grouped[["x_meters", "y_meters"]])

deaths_grouped.head()
```

# Map Chart (v2)

Upgrades the v1 map: deaths are now coloured by the pump they are closest to, and a slider counts deaths around Broad St.

## Rationale:
- v1 showed deaths and pumps but gave no way to measure distance
- Radius OFF: the KNeighborsClassifier tells me which pump each death is closest to
- Radius ON: the slider circles Broad St. and counts the deaths inside the radius

## Conclusion:
- Most deaths are closest to the Broad St. pump
- The slider turns "it looks close" into an actual count and distance
- Still missing: a timeline, and the map uses default styling

```python
#Takes colors from the plotly express library
colors = px.colors.qualitative.Bold

#Maps pump names to corresponding colors
color_map = dict(zip(pumpData["pump_name"], colors))

fig = go.Figure()

#Deaths colored by their nearest pump
fig.add_trace(go.Scattermap(
    lat=deaths_grouped["x_latitude"],
    lon=deaths_grouped["y_longitude"],
    mode="markers",
    name="Deaths",
    customdata=np.c_[deaths_grouped["death_count"], deaths_grouped["nearest_pump"]],
    hovertemplate=("<b>Deaths here:</b> %{customdata[0]}<br>"
                   "<b>Nearest pump:</b> %{customdata[1]}<extra></extra>"),
    marker=dict(size=6 + deaths_grouped["death_count"] * 3,
                color=deaths_grouped["nearest_pump"].map(color_map), opacity=0.5),
    showlegend=False
))

#Pumps
fig.add_trace(go.Scattermap(
    lat=pumpData["x_coordinate"],
    lon=pumpData["y_coordinate"],
    mode="markers+text", name="Pumps",
    text=pumpData["pump_name"], textposition="top center",
    marker=dict(size=16, color=pumpData["pump_name"].map(color_map)),
    showlegend=False
))

#Radius circle around Broad St. (stays empty until the slider is moved)
fig.add_trace(go.Scattermap(lat=[], lon=[],
                            mode="lines", name="Radius",
                            showlegend=False,
                            line=dict(color="crimson", width=3),
                            hoverinfo="skip"))

#Legend entries, one colour per pump
for pump, color in color_map.items():
    fig.add_trace(go.Scattermap(lat=[None], lon=[None],
                                mode="markers", name=pump,
                                marker=dict(size=12, color=color)))

'''----------------------------Radius Slider---------------------------- '''

#Distance of every death from the origin (Broad St.), in meters
dist = np.hypot(deaths_grouped["x_meters"], deaths_grouped["y_meters"])
lat0 = broad["x_coordinate"]
lon0 = broad["y_coordinate"]
angle = np.linspace(0, 2 * np.pi, 100)
#0 = back to the original colours, then 25 m to 300 m in steps of 25
radii = [0] + list(range(25, 301, 25))
original = deaths_grouped["nearest_pump"].map(color_map)

def radius_slider(radius):
    '''
    1. inside is True for every death location within the radius
    2. Deaths inside turn crimson, the rest turn gray
    3. The circle is drawn by converting meters to degrees
    4. The title shows how many deaths are inside
    '''
    inside = dist <= radius
    return go.Frame(
        name=str(radius), traces=[0, 2],
        data=[go.Scattermap(marker=dict(color=np.where(inside, "crimson", "lightgray") if radius else original)),
              go.Scattermap(lat=lat0 + (radius / 111320) * np.sin(angle),
                            lon=lon0 + (radius / (111320 * np.cos(np.radians(lat0)))) * np.cos(angle))],
        layout=go.Layout(title_text=f"Within {radius} m of Broad St.: {deaths_grouped['death_count'][inside].sum()} deaths"
                                    if radius else "Cholera Deaths by Nearest Water Pump: Soho, 1854"))

fig.frames = [radius_slider(radius) for radius in radii]

fig.update_layout(
    title="Cholera Deaths by Nearest Water Pump: Soho, 1854",
    map=dict(style="carto-positron", center=dict(lat=51.5133, lon=-0.1367), zoom=15.5),
    width=1000, height=700,
    legend_title_text="Pumps",
    sliders=[dict(currentvalue=dict(prefix="Radius (m): "), pad=dict(t=30), steps=[
        dict(method="animate", label="Off" if radius == 0 else str(radius),
             args=[[str(radius)], {"frame": {"duration": 0, "redraw": True}, "mode": "immediate"}])
        for radius in radii])]
)
fig.show()
```

# Timeline Chart (v2)

A first timeline of cases and deaths per day.

## Rationale:
- The original map has no time information
- I want to see when the outbreak started, peaked and faded

## Conclusion:
- Cases and deaths both peak in early September
- Hard to tell what happened around the pump removal because nothing on the chart marks it
- Next version: split the outbreak into chapters with a slider and a caption for each

```python
snowDates["date"] = pd.to_datetime(snowDates["date"])
snowDates.rename(columns={'attacks': 'cases'}, inplace=True)

timeline = go.Figure()
timeline.add_trace(go.Scatter(x=snowDates["date"], y=snowDates["cases"],
                              name="Cases per day", mode="lines",
                              line=dict(color='#12436D', width=3)))
timeline.add_trace(go.Scatter(x=snowDates["date"], y=snowDates["deaths"],
                              name="Deaths per day", mode="lines",
                              line=dict(color='#F46A25', width=3)))

timeline.update_layout(
    template="plotly_white",
    title="Cholera Outbreak Timeline: Soho, 1854",
    hovermode="x unified",
    xaxis=dict(tickformat="%d %b", title=None),
    yaxis=dict(rangemode="tozero", title="Per day"),
    width=1000, height=500
)
timeline.show()
```
